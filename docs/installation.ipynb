# INSTALLATION GUIDE


## Installing the dependencies
---------------------------

The sea ice model is pretty much self contained. The only external library needed is [netCDF4](https://www.unidata.ucar.edu/software/netcdf), which itself depends on [HDF5 (>1.8)](https://support.hdfgroup.org/downloads/). On most distributions, HDF5 should be available through the standard repositories. On the other hand, netCDF4 needs to be downloaded and compiled manually. 

### Compiling HDF5

Download the HDF5 source code from [https://support.hdfgroup.org/downloads/](https://support.hdfgroup.org/downloads/).  
Unpack the archive and enter the directory, then:

```
  FC = gfortran configure --enable-fortran --prefix=/usr/local/hdf5
  make
  make check
  sudo make install 
```

### Compiling netCDF4

Download the netCDF4 source code from [https://www.unidata.ucar.edu/software/netcdf](https://www.unidata.ucar.edu/software/netcdf).   
Unpack the archive and enter the directory, then:

```
  export F90=gfortran
  export F77=gfortran
  configure --enable-shared --enable-netcdf-4 --prefix=/usr/local/netcdf4 --with-hdf5=/usr/local/hdf5
  make check
  sudo make install
```

### Linking the libraries
  
To make sure that the system finds the shared netCDF and HDF5 libraries, you will need to add the `lib/` directories to the list of paths searched for  libraries. To do so, go into the `/etc/ld.so.conf.d` directory, and create two files: `netcdf4.conf` and `hdf5.conf`. In each file, write on the first line the path to the libraries, *i.e.* `/usr/local/netcdf4/lib` and `/usr/local/hdf5/lib`, respectively. Then run the command:

```
  sudo /sbin/ldconfig
```

and the linker should be able to find those libraries when you launch the model. 


### Installing scons

The build system is based on [scons](https://scons.org/). If scons is not installed on your system, 
try the following with administrative privileges:

```
  yum install scons
```


## Obtaining the code
----------------------

### clone SIM
Clone SIM from github:

```bash
cd /where/you/want/your/SIM  
git clone git@github.com:McGill-sea-ice/SIM.git  
cd SIM
```

### Forcing
To start, the simplest way is to use the example forcing included in the github repository. For more details on how to construct the forcing based on [ERA5](https://cds.climate.copernicus.eu/datasets/reanalysis-era5-complete?tab=overview) reanalysis see [create_simple_forcing.ipynb](../aux/create_simple_forcing.ipynb) and the [documentation](documentation.ipynb).

In case you do not use the example forcing, you need to link your forcing into the forcing directory `/where/you/want/your/SIM/forcing/`   

### Initial compile
This will compile the code and create the executable `zoupa`.

```bash
./setup_script
export HDF5_DISABLE_VERSION_CHECK=2
```

### Run the code
The `setup_script` linked the inputfile `input` and the namelist `namelist` into the base directory so that the model can find them. The following command starts the simulation with `input` as an inputfile and will store the log messages in `outputfile.log`. To allow for higher resolutions, we compile the code with `mcmodel=medium`, which requires to set `ulimit -s unlimited`, basically allowing the model to access more memory. Note that this setting is kept until you close the bash session.

```bash
ulimit -s unlimited
./zoupa < input > outputfile.log &
```

The model output variables are stored in the `output` directory.  
For subsequent compiles it is sufficient to run `scons-3`.
