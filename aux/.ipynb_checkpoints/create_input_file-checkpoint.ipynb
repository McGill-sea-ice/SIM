# Create inputfiles for SIM
This gives the possbility to easily create a long input file with many "posting dates"

```python
import xarray as xr
```

Define the path where to save the inputfile

```python
path = "/aos/home/jrieck/src/SIM/inputfiles/"
```

Define the parameters to be included in the inputfile

**`exp`**: experiment number  
**`restart`**: `0` for no restart, `1` for restart  
**`restart_date`**: date of the restart file in format `YYYY-MM-DD:hh:mm:ss`  
**`restart_num`**: experiment number of the experiment that created the restart files  
**`namelist`**: `0` for no namelist, `1` for using a namelist  
**`startdate`**: start date of the experiment in format `YYYY-MM-DD:hh:mm:ss`  
**`enddate`**: end date of the experiment in format `YYYY-MM-DD:hh:mm:ss`  
**`startdate_output`**: first date at which to write output (format `YYYY-MM-DD hh:mm:ss`)  
**`enddate_output`**: last date at which to write output (format `YYYY-MM-DD hh:mm:ss`)  
**`frequency_output`**: frequency at which to write output in between `startdate_output` and `enddate_output` (see [`xarray.date_range`](https://docs.xarray.dev/en/stable/generated/xarray.date_range.html#xarray.date_range) for acceptable formats)  

```python
exp = "01"
restart = "0"
if restart == "0":
    isrestart = "norestart"
elif restart == "1":
    isrestart = "restart"
    restart_date = "2008-01-01:00:00:00"
    restart_num = "40"
namelist = "1"
startdate = "2008-01-01:00:00:00"
enddate = "2008-01-31:23:55:00"
startdate_output = "2008-01-01 12:00:00"
enddate_output = "2008-12-31 12:00:00"
frequency_output = "1MS"
od = xr.date_range(start=startdate_output, end=enddate_output, freq=frequency_output)
outputdates = [str(od[i])[0:10] + ":" + str(od[i])[11::] for i in range(0, len(od))]
```

Create a string to write to file from the parameters defined above

```python
if restart == "0":
    inputstring = str(namelist + "                   input namelist\n"
                      + restart + "                   restart\n"
                      + exp + "                  exp version\n"
                      + startdate + " starting date\n"
                      + enddate + " end date\n")
elif restart == "1":
    inputstring = str(namelist + "                   input namelist\n"
                      + restart + "                   restart\n"
                      + restart_date + " year-month-day-hour-minute-second\n"
                      + restart_num + "                  restart version\n" 
                      + exp + "                  exp version\n"
                      + startdate + " starting date\n"
                      + enddate + " end date\n")
for add in range(0, len(outputdates)):
    inputstring = inputstring + outputdates[add] + " posting date\n"
inputstring = inputstring + "stop                end of post date string"
```

Write the inputfile

```python
with open(path + "input_" + isrestart + exp, "w") as file:
    file.write(inputstring)
```
